## What problem does Six Thinking Hats try to solve?

Complex decisions often mix several kinds of thinking at once. One person offers a fact, another attacks the proposal, another defends it, another proposes something new, and another reacts emotionally. The conversation can become an argument about **who is right** instead of a disciplined exploration of **what the situation contains**.

De Bono's method separates thinking into six deliberate modes. Participants metaphorically put on a hat to focus attention on one mode at a time. The method is commonly described as **parallel thinking**: rather than assigning people permanent roles, the group moves through modes together so that everyone can contribute facts, feelings, caution, benefits, creativity, and process thinking.[1][2]

### The hats are not personality types

A person is not "the Black Hat person" or "the creative Green Hat person." The value of the method is that people practice switching modes. The official de Bono Six Thinking Hats site specifically cautions against treating hats as personality types or having meeting participants adopt different hats simultaneously as the authentic group method.[2]

That matters for this repository: **six specialized AI agents are a computational adaptation designed to guarantee perspective coverage.** They are inspired by the hats, but they are not a literal reproduction of how a human Six Hats meeting is normally facilitated.

## The Six Hats Framework

| Hat | Thinking mode | Core questions | Common failure to avoid |
|---|---|---|---|
| **White — Facts** | Data, neutral information, evidence, gaps | What do we know? What is the source? What do we need to learn? | Treating assumptions or model outputs as verified facts |
| **Red — Feelings** | Emotions, intuition, reactions | What is my immediate reaction? How might stakeholders feel? | Disguising feelings as evidence, or dismissing emotion as irrelevant |
| **Black — Caution** | Risks, constraints, logical negatives | What could fail? What conditions make this unsafe or infeasible? | Reflexive pessimism without a mechanism or mitigation |
| **Yellow — Benefits** | Value, strengths, logical positives | What could work? Where is the value? Under what conditions? | Optimism without evidence or measurable benefit |
| **Green — Creativity** | Alternatives, reframing, new ideas | What else could we try? Can we escape a false binary? | Generating novelty without regard to context or feasibility |
| **Blue — Process** | Agenda, sequence, synthesis, next steps | What are we deciding? Which mode is needed now? What happens next? | Confusing process control with having the final substantive answer |

The Blue Hat is especially important in this project. In de Bono's framework it manages the **thinking process**. In our software architecture, a separate **Orchestrator agent** manages the multi-agent workflow. The two are related, but not identical.

## From thinking hats to AI agents

This project turns each thinking mode into a role-specialized AI agent:

```text
                         ORCHESTRATOR
                route • compare • synthesize
             detect gaps • preserve disagreement
                          /  |  \
                         /   |   \
             White  Red  Black  Yellow  Green  Blue
```

Each Hat Agent receives the same problem and shared context but uses a different system prompt and output contract. The Orchestrator then asks:

- Which claims are evidence, assumptions, intuitions, risks, benefits, or alternatives?
- Where do agents agree for **different reasons**?
- Where do they genuinely disagree?
- What important question did no agent ask?
- Are all six agents inheriting the same flawed framing?
- Who is affected but absent from the analysis?
- What are the **elephants in the room**—power, incentives, ethics, maintenance, equity, governance, security, or second-order effects?

The goal is not to hold a six-agent vote. The goal is to construct a more complete model of the problem.

## Why this matters in global health

Global-health decisions are often **multi-country, multi-level, resource-constrained, politically sensitive, evidence-limited, and time-dependent**. A technically correct intervention can still fail if supply chains break, communities distrust it, frontline workflows are unrealistic, incentives are misaligned, or benefits and burdens are distributed inequitably.

A Six Hats lens can create a structured pause before premature convergence:

- **White:** surveillance quality, denominator uncertainty, coverage data, logistics, evidence strength.
- **Red:** fear, trust, stigma, fatigue, morale, community sentiment, political anxiety.
- **Black:** safety, operational failure, inequity, unintended consequences, legal or regulatory constraints.
- **Yellow:** lives saved, resilience, efficiency, workforce benefits, learning, long-term system strengthening.
- **Green:** alternate delivery models, community partnerships, phased pilots, low-resource adaptations, new data pathways.
- **Blue:** decision scope, stakeholder participation, sequencing, criteria, review points, and what must remain provisional.

This complements systems thinking: the hats help **separate modes of attention**, while systems approaches help examine relationships, feedback, delays, and consequences across the larger health system.

## Documented health and global-health uses

The literature contains several useful examples and precedents:

### 1. Global health security and innovation

Stikeleather and Masys include Six Thinking Hats in a chapter on **Global Health Security Innovation**. They describe it as a synthesis-oriented decision process that can help teams work across knowns, known unknowns, unknown unknowns, and unknowables in the volatile and complex global-health-security environment.[3] This is the closest directly documented global-health example in the sources used here.

### 2. Healthcare quality improvement

The NHS *Handbook of Quality and Service Improvement Tools* includes Six Thinking Hats as a structured improvement tool and emphasizes using one hat at a time so participants think in parallel.[4] A 2024 Oxford handbook chapter on preparing healthcare quality-improvement projects also includes Six Thinking Hats among methods for analyzing processes and engaging staff and patients.[5]

### 3. Medical simulation and debriefing

Zhang and colleagues applied Six Thinking Hats to emergency-medicine simulation debriefing. Their feasibility report found that visual hat prompts helped participants remain within the intended thinking frame, and participants reported a non-judgmental environment supportive of discussion.[6]

These examples do **not** prove that Six Thinking Hats improves every health decision. They show that the framework has been adapted to health-security thinking, quality improvement, education, and debriefing—settings where multiple perspectives and disciplined discussion matter.

### Worked example: an outbreak with incomplete information

**Decision:** A neighboring country reports a rapidly growing cluster of severe febrile illness near the border. Should a ministry immediately introduce broad entry screening and movement restrictions?

- **White:** What is the case definition? What is known about transmission? How sensitive is entry screening? How complete are line lists and travel histories?
- **Red:** Communities may be frightened; border workers may feel exposed; travelers may perceive measures as punitive or stigmatizing.
- **Black:** Broad restrictions could divert staff, disrupt livelihoods and supply chains, create informal crossings, and generate false reassurance if screening sensitivity is low.
- **Yellow:** Visible measures could mobilize preparedness, improve cross-border coordination, and create an opportunity to strengthen surveillance and referral.
- **Green:** Consider targeted health messaging, sentinel surveillance, rapid cross-border data sharing, traveler information, risk-based screening, and pre-defined escalation triggers instead of a single all-or-nothing measure.
- **Blue:** Define the decision window, evidence thresholds, accountable authority, neighboring-country coordination, equity criteria, and a date for reassessment.

The **Orchestrator** should then ask what all six hats may still have missed—for example, informal trade routes, migrant populations, laboratory turnaround time, political incentives, or whether the suspected pathogen makes entry screening epidemiologically useful at all.

## The blind-men-and-elephant connection

The parable of the blind men and the elephant is a useful companion metaphor for this project. Each observer can have a locally valid experience and still hold an incomplete model of the whole.

Six Hat Agents are intentionally **partial observers**. Their specialization is useful precisely because it is limited. The Orchestrator's responsibility is therefore not to select the most persuasive hat, but to integrate partial views, preserve uncertainty, challenge shared assumptions, and identify what lies outside the current frame.

> **Design principle:** A multi-agent system becomes more useful when specialization creates productive differences **and** orchestration prevents those differences from becoming six isolated blind spots.
