# Lab 1 — Build and run the Six Thinking Hats agents

This notebook runs entirely in your browser in JupyterLite. You will inspect six role prompts, send the same problem to each role, and compare the resulting perspectives.

The default `TeachingBackend` is deterministic and uses no network calls or API keys. Its purpose is to make the architecture visible. Later you can replace it with a real model backend without changing the hat agents.

```python
from six_hats_agents import HATS, HatAgent, TeachingBackend, print_perspectives

backend = TeachingBackend()
for name, config in HATS.items():
    print(f"{config.label}: {config.purpose}")
```

## Inspect one agent prompt

An agent is defined partly by its role prompt. Notice that the White Hat receives both role-specific instructions and shared rules.

```python
white = HatAgent(HATS["white"], backend)
system_prompt, user_prompt = white.build_prompts(
    "Should our organization pilot an AI assistant for a repetitive internal workflow?",
    "The pilot must be low-cost, reversible, and measurable."
)
print("SYSTEM PROMPT\n", system_prompt)
print("\nUSER PROMPT\n", user_prompt)
```

## Give all six agents the same problem

Edit the strings below and re-run the cell. The hats should not vote; they should expose different kinds of information.

```python
problem = "Should our organization pilot an AI assistant for a repetitive internal workflow?"
context = "The pilot must be low-cost, reversible, measurable, and protect sensitive information."

perspectives = {
    name: HatAgent(config, backend).analyze(problem, context)
    for name, config in HATS.items()
}
print_perspectives(perspectives)
```
