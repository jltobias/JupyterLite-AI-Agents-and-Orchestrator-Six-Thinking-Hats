# Lab 2 — Orchestrate the six perspectives and look for the whole elephant

The orchestrator sits outside the Six Hats framework. It calls each hat, preserves the outputs, compares them, and creates a synthesis that explicitly searches for blind spots and “elephants in the room.”

```python
from six_hats_agents import SixHatsOrchestrator, print_perspectives

orchestrator = SixHatsOrchestrator()
```

## Run a complete deliberation

Replace the example with a problem you care about. Give all hats the same context so differences are due to perspective rather than hidden inputs.

```python
problem = "Should a city replace some scheduled inspections with risk-based inspections supported by analytics?"
context = "The city wants faster service without reducing safety, fairness, transparency, or public trust."

result = orchestrator.run(problem, context)
print_perspectives(result["perspectives"])
```

## Inspect the synthesis

The teaching orchestrator is intentionally transparent. It does not claim to infer deep semantic truth from the six responses. Instead, it demonstrates the categories a real LLM orchestrator should preserve and challenge.

```python
print(result["synthesis"])
```

## Blind men and elephant exercise

For each hat, write one sentence describing the “part of the elephant” it sees most clearly. Then ask a more difficult question: **what assumption might all six hats share because it came from the original problem framing?**

Try changing the problem from a yes/no question into a design question. For example:

> Instead of “Should we adopt X?”, ask “What combination of process, technology, safeguards, and evidence would improve outcome Y?”

Re-run the orchestrator and compare how the hats change.

```python
reframed_problem = "What combination of inspection scheduling, analytics, human review, safeguards, and evaluation would improve service speed while preserving safety, fairness, transparency, and trust?"
reframed = orchestrator.run(reframed_problem, context)
print(reframed["synthesis"])
```

## Optional second-pass design

A stronger production workflow could:

1. run all six hats independently;
2. create a first synthesis;
3. ask White to flag unsupported claims;
4. ask Black to stress-test the proposed next step;
5. ask Green for an alternative that avoids the largest risk;
6. ask Blue whether the process has enough evidence to decide;
7. revise the synthesis and record what changed.

This transforms the workflow from simple fan-out/fan-in into an iterative deliberation loop.
